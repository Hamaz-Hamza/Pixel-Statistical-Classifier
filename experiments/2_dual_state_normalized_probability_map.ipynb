```python
from keras.datasets import mnist
import matplotlib.pyplot as plt
import numpy as np
from psc.models.dual_state_normalized_probability_map import PixelStatisticalClassifier
from psc import evaluation
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
for foreground_pixel_weight in [0, 0.25, 0.5, 0.75, 1]:
    print("foreground pixel weight:", foreground_pixel_weight)
    psc = PixelStatisticalClassifier(foreground_pixel_weight)
    psc.fit(x_train, y_train)
    results = evaluation.evaluate(psc, x_test, y_test)
    evaluation.display_results(results)
    print("-" * 40)
```

```python
# <<< Analysis >>>
#
# results have improved drastically, indicating that the previous model was biased towards majority background pixel predictions.
# as seen above, foreground pixels are highly powerfull discriminators for the classes
# as when only background pixels voted only 11.35% overall accuracy was achieved compared to 65.63% when only foreground pixels voted. 
# high precision, low recall scores for digits 4 and 5 indicate that those were rarely predicted 
# whereas the opposite case for digit 0 shows that the model is biased towards the 0 digit
# high accuracies and f1 scores for digits 1, 2, 6, and 7 indicate that the model is great at predicting those cases
#
# these results indicate that using the probabilistic model counterattacks the negative effects of
# not only state imbalances (background dominating over foreground or viceversa)
# but also statistical imbalances (10x more samples for background than for foreground or viceversa)

# <<<  Next step >>>
#
# Instead of frequency aggregation, learn probabilities for each class based on state for each pixel
# Then, during prediction, average the probabilities across pixels with the same states
# Finally, use a tunable parameter to choose the weight assigned to the states
```

```python
import mnists
from psc.models.pixel_statistical_classifier import PixelStatisticalClassifier
from psc.evaluation import evaluate_on_dataset, evaluate, display_results
```

```python
mnist_dataset = mnists.MNIST("../datasets/mnist")
evaluate_on_dataset(mnist_dataset, PixelStatisticalClassifier())
```

```python
fashion_mnist_dataset = mnists.FashionMNIST("../datasets/fashion_mnist")
evaluate_on_dataset(fashion_mnist_dataset, PixelStatisticalClassifier())
```

```python
kmnist_dataset = mnists.KMNIST("../datasets/kmnist")
evaluate_on_dataset(kmnist_dataset, PixelStatisticalClassifier())
```

```python
# manual download is required
emnist_letters_dataset = mnists.EMNIST("../datasets/emnist").Letters()
evaluate_on_dataset(emnist_letters_dataset, PixelStatisticalClassifier())
```

```python
# manual download is required
emnist_digits_dataset = mnists.EMNIST("../datasets/emnist").Digits()
evaluate_on_dataset(emnist_digits_dataset, PixelStatisticalClassifier())
```

```python
x_train = emnist_digits_dataset.train_images()
y_train = emnist_digits_dataset.train_labels()
x_test = emnist_digits_dataset.test_images()
y_test = emnist_digits_dataset.test_labels()

model = PixelStatisticalClassifier()
```

```python
model.fit(x_train[:1000], y_train[:1000])
results = evaluate(model, x_test, y_test)
display_results(results)
```

```python
model.fit(x_train[:2000], y_train[:2000])
results = evaluate(model, x_test, y_test)
display_results(results)
```
